# Causal feature correspondence: when similarity lies

This v0.8 tutorial compares two source features that both predict the same target representation extremely well. One feature is genuinely used by the source computation; the other is a correlated decoy with the same semantic label. The scientific question is whether held-out intervention transfer can tell them apart.

## Evidence layers

Keep these separate: activation correlation, geometry, semantic labels, predictive transfer, intervention-effect agreement, and causal substitution. A high score in the first four categories is not permission to call two features mechanistically corresponding.

```python
from neuros_mechint.benchmarks.correspondence_ground_truth import run_correspondence_ground_truth_benchmark

report = run_correspondence_ground_truth_benchmark(seed=0)
report.to_dict()
```

## The trap

The decoy is constructed to have nearly perfect held-out predictive R² and the same semantic label as the target. If we stopped at representation similarity, we would promote the wrong feature. Its source-ablation effect is zero, so v0.8 gives it no causal credit.

```python
assert report.true_correspondence_passed
assert report.decoy_similarity_high
assert report.decoy_causal_rejected
assert report.decoy_validation_predictive_r2 > 0.99
assert report.decoy_median_source_effect == 0.0
```

## What the true candidate must survive

1. Fit the mapping on discovery trials only.
2. Predict target activations on content-distinct held-out trials.
3. Show the source feature is causally relevant under ablation.
4. Show the target feature is causally relevant under ablation.
5. Restore the target contribution using the frozen source-to-target map.
6. Beat shuffled semantic-trial donors.
7. Beat same-cardinality random source feature sets, each of which receives its own discovery-only mapping.

```python
print('held-out R2:', report.true_validation_predictive_r2)
print('median causal recovery:', report.true_median_causal_recovery)
print('margin over shuffled pairs:', report.true_shuffled_margin)
print('random-control percentile:', report.true_random_percentile)
```

## Taking this to real neural foundation models

Use `FeatureSpaceIdentity` to pin the source and target representations, `AdapterFeatureSpaceView` to expose a model component, and `run_adapter_feature_correspondence_study(...)` for real ablation and substitution. The default tensor projector averages non-feature axes; temporal or event-specific claims should use a projector that preserves those coordinates.

A useful next real study is to take an estimable v0.7 Transformer × SSM / Event × Relative-ISI interaction and ask whether the affected conditions recruit genuinely different causal features or merely different bases for a shared computation.

## Claim boundary

A passing held-out substitution study establishes conditional causal substitutability under the frozen models, mapping, projector, metric, trials, and intervention controls. It does not establish biological homology, feature uniqueness, universal semantic identity, or in-manifold intervention validity.
