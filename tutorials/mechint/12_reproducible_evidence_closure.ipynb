# 12 · Reproducible evidence closure

v1 separates two questions that are easy to blur: **is the software/evidence contract reproducible?** and **have the real neural experiments actually been completed?** This lesson exercises the first while refusing to invent the second.

```python
from neuros_mechint import default_v1_evidence_status, schema_catalog
from neuros_mechint.benchmarks import (
    ReproductionMetricTolerance, ReproductionSnapshot, ReproductionSpec,
    assess_independent_reproduction,
)

print('schema families:', [item['family'] for item in schema_catalog()])
status = default_v1_evidence_status()
print('software contract ready:', status.software_contract_ready)
print('empirical evidence complete:', status.empirical_evidence_complete)
print('pending empirical requirements:', len(status.pending_empirical_requirements))
```

## Independent execution is not byte identity

Two executions should share a deterministic scientific fingerprint but have different run identities. Their qualitative scientific decision must agree, and preregistered numerical metrics must remain within tolerance.

```python
spec = ReproductionSpec(
    reproduction_id='tutorial-rerun', artifact_family='correspondence',
    required_decision='promoted',
    metric_tolerances=(ReproductionMetricTolerance('causal_recovery', absolute=0.02, relative=0.0),),
)
reference = ReproductionSnapshot(
    'correspondence', 'scientific-study', 'run-a', 'machine-a', 'promoted',
    {'causal_recovery': 0.80},
)
rerun = ReproductionSnapshot(
    'correspondence', 'scientific-study', 'run-b', 'machine-b', 'promoted',
    {'causal_recovery': 0.81},
)
result = assess_independent_reproduction(spec, reference, rerun)
print(result.to_dict())
assert result.passed
```

## The anti-overclaim check

The v1 package may be software-contract ready while the repository still reports the real Transformer × SSM × Event × Relative-ISI evidence program as pending. That is intentional. A framework should make it easier to say **not yet measured** than to accidentally upgrade a synthetic benchmark into a neuroscience result.
