## The evidence ladder

A mechanistic claim becomes stronger as we move from observing a correlation to intervening on the proposed mechanism and attempting to falsify it.

1. **Localization:** where does a signal appear?
2. **Necessity:** does removing the proposed mechanism hurt the target behavior?
3. **Sufficiency:** does restoring or retaining it recover the behavior?
4. **Specificity:** does the intervention affect what we predicted more than unrelated behavior?
5. **Stability:** does the result survive seeds, examples, checkpoints, and reasonable method choices?
6. **Held-out validity:** was the mechanism evaluated on data that did not select it?
